## ACCESSCORP STOCK PREDICTION

```python
import pandas as pd

accesscorp=pd.read_csv('../stock_data/ACCESSCORP.csv')
accesscorp
```

```python
accesscorp=accesscorp.drop(columns=['high_price','low_price'])
```

```python
accesscorp.rename(columns={'trade_date':'date'},inplace=True)
```

```python
accesscorp['date']=pd.to_datetime(accesscorp['date'])
accesscorp.info()
```

```python
accesscorp['month']=accesscorp['date'].dt.month
accesscorp['quarter']=accesscorp['date'].dt.quarter
accesscorp['year']=accesscorp['date'].dt.year
```

```python
import matplotlib.pyplot as plt

plt.plot(accesscorp['close_price'])
plt.show()
```

```python
accesscorp_end_month_df = (
    accesscorp
    .sort_values("date")
    .groupby(accesscorp["date"].dt.to_period("M"))
    .tail(1)
    .reset_index(drop=True)
)

accesscorp_end_month_df
```

```python
plt.plot(accesscorp_end_month_df['close_price'])
plt.show()
```
