```python
## to change the name across the code
## ctrl + h will open the tab the fill the new name in replace
## to change the name across the entire code so you dont have to click enter everytime
## ctrl + shift + h

import pandas as pd

cadbury=pd.read_csv('../stock_data/cadbury.csv') ## so it can be small letter. This is cool.
cadbury=cadbury.drop(columns=['high_price','low_price'])
cadbury.rename(columns={'trade_date':'date'},inplace=True)
cadbury['date']=pd.to_datetime(cadbury['date'])
cadbury['month']=cadbury['date'].dt.month
cadbury['quarter']=cadbury['date'].dt.quarter
cadbury['year']=cadbury['date'].dt.year

cadbury
```

```python
import matplotlib.pyplot as plt

plt.plot(cadbury['close_price'])
plt.show()
```

```python
cadbury_end_month_df = (
    cadbury
    .sort_values("date")
    .groupby(cadbury["date"].dt.to_period("M"))
    .tail(1)
    .reset_index(drop=True)
)

cadbury_end_month_df
```

```python
plt.plot(cadbury_end_month_df['close_price'])
plt.show()
```

```python
cadbury_end_month_df['returns']=cadbury_end_month_df['close_price'].pct_change()*100
cadbury_end_month_df
```

```python
plt.plot(cadbury_end_month_df['returns'])
plt.show()
```

```python
## Test stationarity

from statsmodels.tsa.stattools import adfuller

result = adfuller(cadbury_end_month_df['returns'].dropna())
print(result)
pvalue = result[1]
if pvalue < 0.05:
    print('stationary')
else:
    print('not stationary')
```

```python
# Plotting the autocorrelation function and the partial autocorrelation function
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
plot_acf(cadbury_end_month_df['returns'].dropna(), lags=10)
plot_pacf(cadbury_end_month_df['returns'].dropna(), lags=10)
plt.show()
```
