## Stock data feature engineering
### 1. Derive the returns in Percentage
This is the percentage difference between the current closing price and previous price 

```python
aiico_data_2['returns_pct']=aiico_data_2['close_price'].pct_change()*100
aiico_data_2
```

```python
aiico_data_2.drop(['returns'],axis=1,inplace=True)
```

```python
aiico_data_2
```

### 2. Introduce Lag returns
Lagging is when you include previous information into current data point to give a better information about the past and how it affects the present and future

We will apply Lagging on the returns of previous return(lag1), previous 5 return(lag5), previous 10 return(lag10)

```python
aiico_data_2["lag_return1"] = aiico_data_2["returns_pct"].shift(1)
aiico_data_2["lag_return2"] = aiico_data_2["returns_pct"].shift(2)
aiico_data_2["lag_return5"] = aiico_data_2["returns_pct"].shift(5)
aiico_data_2["lag_return10"] = aiico_data_2["returns_pct"].shift(10)
```

```python
aiico_data_2.head(15)
```

### 3. Applying Moving Averages

Instead of looking at one day's price, we calculate the average of recent prices. This will smoothen the plot to give a sense of the trend. The more the averaging gap, the better the smoothening.

* 5-day	    Very short-term movement
* 10-day	Short-term trend
* 20-day	About one trading month
* 50-day	Medium-term trend

We will use 5(MA_5), 20(MA_20) and 50(MA_50) for this exercise.

```python
aiico_data_2.shape
```

```python
aiico_data_2["stock_MA_5"] = aiico_data_2["close_price"].rolling(window=5).mean()
aiico_data_2["stock_MA_20"] = aiico_data_2["close_price"].rolling(window=20).mean()
aiico_data_2["stock_MA_50"] = aiico_data_2["close_price"].rolling(window=50).mean()
```

```python
aiico_data_2.head(20)
```

### Apply Rolling Volatility
Volatility measures how much returns fluctuate over a period. If i pick a returns period of 20 days , I use this to find out how stable the returns is over 20 days. This is basically the standard deviation during the period.

We will be using Rolling volatility of 20 days

```python
aiico_data_2["returns_volatility_20"] = (
    aiico_data_2["returns_pct"]
      .rolling(window=20)
      .std()
)
```
