## Stock data feature engineering
### 1. Derive the returns in Percentage
This is the percentage difference between the current closing price and previous price 

```python
aiico_data_2['returns_pct']=aiico_data_2['close_price'].pct_change()*100
aiico_data_2
```

```python
aiico_data_2
```

### 2. Introduce Lag returns
Lagging is when you include previous information into current data point to give a better information about the past and how it affects the present and future

We will apply Lagging on the returns of previous return(lag1), previous 5 return(lag5), previous 10 return(lag10)

```python
aiico_data_2["lag_return1"] = aiico_data_2["returns_pct"].shift(1)
aiico_data_2["lag_return2"] = aiico_data_2["returns_pct"].shift(2)
aiico_data_2["lag_return5"] = aiico_data_2["returns_pct"].shift(5)
aiico_data_2["lag_return10"] = aiico_data_2["returns_pct"].shift(10)
```

```python
aiico_data_2.head(15)
```

### 3. Applying Moving Averages

Instead of looking at one day's price, we calculate the average of recent prices. This will smoothen the plot to give a sense of the trend. The more the averaging gap, the better the smoothening.

* 5-day	    Very short-term movement
* 10-day	Short-term trend
* 20-day	About one trading month
* 50-day	Medium-term trend

We will use 5(MA_5), 20(MA_20) and 50(MA_50) for this exercise.

```python
aiico_data_2.shape
```

```python
aiico_data_2["stock_MA_5"] = aiico_data_2["close_price"].rolling(window=5).mean()
aiico_data_2["stock_MA_20"] = aiico_data_2["close_price"].rolling(window=20).mean()
aiico_data_2["stock_MA_50"] = aiico_data_2["close_price"].rolling(window=50).mean()
```

### 4. Apply Rolling Volatility
Volatility measures how much returns fluctuate over a period. If i pick a returns period of 20 days , I use this to find out how stable the returns is over 20 days. This is basically the standard deviation during the period.

We will be using Rolling volatility of 20 days

```python
aiico_data_2["returns_volatility_20"] = (
    aiico_data_2["returns_pct"]
      .rolling(window=20)
      .std()
)

aiico_data_2["returns_volatility_50"] = (
    aiico_data_2["returns_pct"]
      .rolling(window=50)
      .std()
)
```

### 6. MACD

MACD stands for Moving Average Convergence Divergence.

It is a trend-following momentum indicator.

It answers the question: `"Is the stock's short-term trend becoming stronger or weaker compared to its longer-term trend?"`

Unlike RSI, which measures the strength of buying and selling, MACD measures changes in momentum.


MACD uses `Exponential Moving Averages (EMA)` instead of simple moving averages.

An EMA gives more weight to recent prices, making it more responsive to new information.

The standard settings are:

- 12-day EMA (short-term)
- 26-day EMA (long-term)

`MACD=EMA12​−EMA26​`

Positive MACD means

`The short-term average is above the long-term average.`

This usually indicates upward momentum.


Negative MACD means

`The short-term trend is weaker than the long-term trend.`

Often interpreted as bearish momentum.



The Signal Line is simply
`A 9-day EMA of the MACD line`

So now we have two lines:

- MACD Line
- Signal Line


If The MACD crosses above the Signal Line.
Many traders interpret this as increasing upward momentum.

Now If the MACD crosses below the Signal Line.
This may indicate weakening momentum.


* **The histogram** is simply
`Histogram=MACD−Signal`


Positive bars suggest bullish momentum is stronger than the signal line.
If the histogram becomes negative, momentum has weakened relative to the signal.


We will use pandas_ta inbuilt formula for these calculations. It will automaticallty create the following columns

* MACD_12_26_9: MACD
* MACDs_12_26_9: SIGNAL line
* MACDh_12_26_9: HISTOGRAM

### Automate the Process

Based on the steps taken we have to find a way to smartly automate this process for all stock data.

```python
import pandas as pd
import pandas_ta as ta
import numpy as np
import os

import pandas as pd

def prepare_data(data: pd.DataFrame,drop_items: list[str] | None = None) -> pd.DataFrame:
    if drop_items is None:
        drop_items = ["high_price", "low_price"]
    return data.drop(columns=drop_items)

def stock_returns_percent(data: pd.DataFrame) -> pd.DataFrame:
    data['returns_pct']=data['close_price'].pct_change()*100
    return data

def lag_returns(data: pd.DataFrame,lag_items: list[int] | None = None) -> pd.DataFrame:
    if lag_items is None:
        lag_items=[1]
    for lag in lag_items:
        data[f'lag_return{lag}']=data['returns_pct'].shift(lag)
    return data
    
def moving_averages(data: pd.DataFrame,averaging_items: list[int] | None = None) -> pd.DataFrame:
    if averaging_items is None:
        averaging_items=[5]
    for avg in averaging_items:
        data[f'stock_MA_{avg}']=data['close_price'].rolling(avg).mean()
    return data

def rolling_volatility(data: pd.DataFrame,volatility_items: list[int] | None = None) -> pd.DataFrame:
    if volatility_items is None:
        volatility_items=[20]
    for vol in volatility_items:
        data[f'returns_volatility_{vol}']=data['returns_pct'].rolling(vol).std()
    return data

def relative_strength_index(data:pd.DataFrame, rsi_length:int | None=None) -> pd.DataFrame:
     if rsi_length is None:
        rsi_length=14

     data['stock_RSI']=ta.rsi(data['close_price'], length=rsi_length)
     return data

def ma_convergence_divergence(data:pd.DataFrame) -> pd.DataFrame:
     data.ta.macd(close='close_price', append=True)
     data.rename(columns={'MACD_12_26_9':'MACD','MACDh_12_26_9':'MACD_hist','MACDs_12_26_9':'MACD_signal'},inplace=True)
     return data

def volume_moving_average(data: pd.DataFrame,volume_items: list[int] | None = None) -> pd.DataFrame:
    if volume_items is None:
        volume_items=[20]
    for vol in volume_items:
        data[f'stock_volume_MA_{vol}']=data['volume'].rolling(vol).mean()
    return data

def clean_and_save(data:pd.DataFrame,stock_name:str):
     cleaned_data=data.dropna()
     cleaned_data.to_csv(f'feature_engineering/stocks_FE/{stock_name.upper()}_ENGINEERED',index=False)


def stock_data_engineering(folder_path):
    for csv_data in os.listdir(folder_path):
        try:
            if csv_data.endswith(".csv"):
                data = pd.read_csv(os.path.join(folder_path, csv_data))

            if len(data) > 400:
                ticker = os.path.splitext(csv_data)[0] ## Captures the name of the stock
                print()
                print(f'{csv_data} engineering begins')

                ## Prepare the data
                print("preparing data")
                prepared_data=prepare_data(data,drop_items=["high_price", "low_price"])

                ## Implement stock returns (%)
                print("stock returns")
                data_with_stock_returns=stock_returns_percent(prepared_data)

                ## Implement lagging
                print("lag returns")
                data_lag_returns=lag_returns(data_with_stock_returns,lag_items=[1,5,10])

                ## Applying Moving Averege
                print("moving_averages")
                data_moving_average=moving_averages(data_lag_returns,averaging_items=[5,20,50])

                ## Applying Rolling Volatility
                print("rolling_volatility")
                data_rolling_volatility = rolling_volatility(data_moving_average,volatility_items=[20,50])

                ## Intrducing RSI
                print("relative_strength_index")
                data_rsi=relative_strength_index(data_rolling_volatility,rsi_length=14)

                ## Calculating the MACD
                print("ma_convergence_divergence")
                data_macd=ma_convergence_divergence(data_rsi)

                ## Get stock volume moving average
                print("volume_moving_average")
                data_volume_ma = volume_moving_average(data_macd,volume_items=[20])

                ## Clean and save data
                print("clean_and_save")
                clean_and_save(data_volume_ma,stock_name=ticker)

                print(f"{ticker} engineering completed")
            else:
                continue
            
        except Exception as e:
            print(f"Error! {ticker}", e)

       
```

## Macroeconomic Data Engineering

We will create a function that will make us choose the number of lagging required for any variable. We will keep maximum lag column for each feature at 3 ie. oil_price can only have lag oil_price_lag(n) three times. We might not need this at all. We will just create this function for future use (reusability)

```python
import pandas as pd
macro_data=pd.read_csv('macroeconomic_data/populated_cleaned_macro_data')
```

```python
import pandas as pd

def daily_lag(data: pd.DataFrame,feature: str,lag_list: list[int] | None = None) -> pd.DataFrame:
    """
    Create lagged versions of a daily feature.
    Maximum of three lag features is advisable.
    NOTE: Drop NaN after lagging.
    """
    if lag_list is None:
        lag_list = [1]

    for lag in lag_list:
        if lag > 0:
            data[f"{feature}_lag{lag}"] = data[feature].shift(lag)

    return data


def monthly_lag(data: pd.DataFrame,feature: str,lag_list: list[int] | None = None) -> pd.DataFrame:
    '''Monthly Lag function. It is advisable to keep at maximum lag instance of 3'''
    
    if lag_list is None:
        lag_list = [1]

    monthly_data = (
        data.groupby(['Year', 'Month_idx'])[feature]
        .first()
        .reset_index()
        .sort_values(['Year', 'Month_idx'])
    )

    for lag in lag_list:
        monthly_data[f'{feature}_lag{lag}'] = monthly_data[feature].shift(lag)

    data = data.merge(
        monthly_data.drop(columns=[feature]),
        on=['Year', 'Month_idx'],
        how='left'
    )

    return data


def quarterly_lag(data: pd.DataFrame,feature: str,lag_list: list[int] | None = None) -> pd.DataFrame:
    '''Quarterly Lag function. It is advisable to keep at maximum lag instance of 3'''
    
    if lag_list is None:
        lag_list = [1]

    quarterly_data = (
        data.groupby(['Year', 'Quarter'])[feature]
        .first()
        .reset_index()
        .sort_values(['Year', 'Quarter'])
    )

    for lag in lag_list:
        quarterly_data[f'{feature}_lag{lag}'] = quarterly_data[feature].shift(lag)

    print(data.columns.tolist())
    print(quarterly_data.columns.tolist())

    data = data.merge(
        quarterly_data.drop(columns=[feature]),
        on=['Year', 'Quarter'],
        how='left'
    )

    return data
```

```python
def lag_features(data: pd.DataFrame,features_lag: dict[str, list[int]] | None = None) -> pd.DataFrame:
    '''Creates lag features based on the frequency suffix.'''

    if features_lag is None:
        return data

    for feature, lag_list in features_lag.items():

        if feature.endswith('_daily'):
            data = daily_lag(
                data=data,
                feature=feature,
                lag_list=lag_list
            )

        elif feature.endswith('_monthly'):
            data = monthly_lag(
                data=data,
                feature=feature,
                lag_list=lag_list
            )

        elif feature.endswith('_quarterly'):
            data = quarterly_lag(
                data=data,
                feature=feature,
                lag_list=lag_list
            )

    return data
```
