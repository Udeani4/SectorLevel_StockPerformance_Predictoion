```python
mde2['FR_Date'] = pd.to_datetime(mde2['FR_Date']).dt.strftime('%Y-%m-%d')
```

```python
mde2["date"] = pd.date_range(
    start="2017-01-31",
    periods=len(mde2),
    freq="ME"
)
```

```python
mde2
```

shift the foreign reserve data to avoid look ahead bias we will shift the monthly data by 1 month. So the information is used to predict the next months returns.

```python
mde2[['foreign_reserve_Gross(USD)_monthly','foreign_reserve_Liquid(USD)_monthly']]=mde2[['foreign_reserve_Gross(USD)_monthly','foreign_reserve_Liquid(USD)_monthly']].shift(1)
```

```python
mde2
```

```python
mde3=mde2.drop(columns=['year','month','month_idx'])
```

```python
mde3
```

```python
for column in mde3.columns:
    if column.endswith('_monthly'):
        mde3[column]=mde3[column].shift(1)
```

```python
## We performed the shift in mde2
mde3
```

```python
mde3[['foreign_reserve_Gross(USD)_monthly','foreign_reserve_Liquid(USD)_monthly']]=mde3[['foreign_reserve_Gross(USD)_monthly','foreign_reserve_Liquid(USD)_monthly']].shift(-1)
```

```python
mde3
```

We are now done with the macroeconomic variables. We will head over to the stock_data. We will start with the finance sector stock data.

```python
fin_sec_data=pd.read_csv('feature_engineering/stocks_sectors/financial_services/financial_services_sector')
```

```python
fin_sec_data
```

```python
fin_sec_data_clean=fin_sec_data.copy()
```

```python
fin_sec_data_clean['sector_volume']=fin_sec_data_clean['sector_volume'].shift(1)
```

```python
fin_sec_data_clean=fin_sec_data_clean.drop('sector',axis=1)
```

```python
fin_sec_data_clean
```

```python
fin_sec_data_clean=fin_sec_data_clean.dropna()
```

```python
fin_sec_data_clean['date']=pd.to_datetime(fin_sec_data_clean['date'])
```

```python
print(fin_sec_data_clean['date'].dtypes)
```

```python
mde3["month"] = mde3["date"].dt.to_period("M")
fin_sec_data_clean["month"] = fin_sec_data_clean["date"].dt.to_period("M")
```

```python
mde3
```

```python
fin_sec_data_clean
```

```python
## Get last day of the month
fin_sec_monthly = (
    fin_sec_data_clean
    .sort_values("date")
    .groupby("month")
    .last()
    .reset_index()
)
```

```python
fin_sec_monthly
```

```python
## We will merge based on the month
fin_sector_merged = mde3.merge(
    fin_sec_monthly.drop(columns="date"),
    on="month",
    how="left"
)
```

```python
fin_sector_merged
```

```python
fin_sector_merged=fin_sector_merged.dropna()
```

```python
fin_sector_merged
```

```python
## remove unuseful columns
fin_sector_merged_clean=fin_sector_merged.drop(columns=['date','month'])
```

```python
fin_sector_merged_clean
```
