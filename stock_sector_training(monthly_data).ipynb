```python
import pandas as pd
from datetime import datetime as dt

# 1. Generate the date range from January 1, 2017, up to today
date_series = pd.date_range(start='2017-01-01', end=dt.now(), freq='ME')

# 2. Create the DataFrame and format the dates as %Y-%m-%d
df = pd.DataFrame()
df['date'] = date_series.strftime('%Y-%m-%d')
```
